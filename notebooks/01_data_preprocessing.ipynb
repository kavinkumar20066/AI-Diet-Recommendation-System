# 9 Cleaning/preprocessing
Dropping duplicates and ensuring numeric columns for features and target.

```python
features = ['Fat', 'Carbohydrates', 'Sugars', 'Protein', 'Dietary Fiber']
target = 'Caloric Value'
required = features + [target]

df_clean = df.drop_duplicates()
df_clean = df_clean.dropna(subset=required)

for col in required:
    df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')

df_clean = df_clean.dropna(subset=required)
```
